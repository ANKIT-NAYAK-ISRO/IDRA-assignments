# Processed E-commerce Dataset Using Python and Pandas

## Objective
The objective of this assignment is to create a processed e-commerce dataset using three CSV files: Orders, Customers, and Products. The datasets are loaded using Pandas and related information is combined using `merge()`. The `concat()` function is demonstrated for combining DataFrames. The `apply()` function is used to create useful transformed columns, while DateTime operations are used to extract information such as month, day, and day of the week. Finally, a clean and meaningful processed dataset is exported as a CSV file.


## 1. Import Required Libraries

```python
import pandas as pd
```

## 2. Load the Input Datasets

```python
orders = pd.read_csv("Day9_Orders.csv")
products = pd.read_csv("Day9_Products.csv")
customers = pd.read_csv("Day9_Customers.csv")

print("Orders Shape:", orders.shape)
print("Products Shape:", products.shape)
print("Customers Shape:", customers.shape)
```

## 3. Display the First Few Rows of Each Dataset

```python
print("Orders Dataset")
display(orders.head())

print("Products Dataset")
display(products.head())

print("Customers Dataset")
display(customers.head())
```

## 4. Check Dataset Information

```python
print("Orders Information:")
orders.info()

print("\nProducts Information:")
products.info()

print("\nCustomers Information:")
customers.info()
```

## 5. Demonstration of concat()

```python
# Combining the first 3 and last 3 rows of the Orders DataFrame
concat_demo = pd.concat(
    [orders.head(3), orders.tail(3)],
    ignore_index=True
)

concat_demo
```

## 6. Convert Order Date to DateTime Format

```python
orders["Order_Date"] = pd.to_datetime(orders["Order_Date"])

orders.info()
```

## 7. Merge Orders with Customers

```python
orders_customers = orders.merge(
    customers,
    on="Customer_ID",
    how="left"
)

orders_customers.head()
```

## 8. Merge the Combined Data with Products

```python
merged_df = orders_customers.merge(
    products,
    on="Product_ID",
    how="left"
)

merged_df.head()
```

## 9. Create a Total Amount Column

```python
merged_df["Total_Amount"] = (
    merged_df["Quantity"] * merged_df["Unit_Price"]
)

merged_df[[
    "Order_ID",
    "Product_Name",
    "Quantity",
    "Unit_Price",
    "Total_Amount"
]].head()
```

## 10. Use DateTime Operations to Extract Date Information

```python
merged_df["Order_Month"] = merged_df["Order_Date"].dt.month
merged_df["Order_Day"] = merged_df["Order_Date"].dt.day
merged_df["Day_of_Week"] = merged_df["Order_Date"].dt.day_name()

merged_df[[
    "Order_ID",
    "Order_Date",
    "Order_Month",
    "Order_Day",
    "Day_of_Week"
]].head()
```

## 11. Use apply() to Create Useful Columns

```python
# Classify each order based on its total amount
merged_df["Order_Size"] = merged_df["Total_Amount"].apply(
    lambda x: "High Value"
    if x >= 5000
    else ("Medium Value" if x >= 2000 else "Low Value")
)

# Transform membership information
merged_df["Customer_Category"] = merged_df["Membership_Type"].apply(
    lambda x: f"{x} Member"
)

merged_df[[
    "Order_ID",
    "Total_Amount",
    "Order_Size",
    "Membership_Type",
    "Customer_Category"
]].head()
```

## 12. Create the Final Processed Dataset

```python
final_columns = [
    "Order_ID",
    "Order_Date",
    "Customer_ID",
    "Customer_Name",
    "City",
    "Region",
    "Membership_Type",
    "Customer_Category",
    "Product_ID",
    "Product_Name",
    "Category",
    "Brand",
    "Unit_Price",
    "Quantity",
    "Total_Amount",
    "Payment_Method",
    "Order_Status",
    "Order_Month",
    "Order_Day",
    "Day_of_Week",
    "Order_Size"
]

final_df = merged_df[final_columns]

# Sort the final dataset by Order Date and Order ID
final_df = final_df.sort_values(
    by=["Order_Date", "Order_ID"]
).reset_index(drop=True)

final_df.head()
```

## 13. Check Missing Values and Dataset Structure

```python
print("Final Dataset Shape:", final_df.shape)

print("\nMissing Values:")
print(final_df.isnull().sum())

print("\nDataset Information:")
final_df.info()
```

## 14. Export the Final Processed Dataset

```python
final_df.to_csv(
    "Processed_Ecommerce_Dataset.csv",
    index=False
)

print("Processed dataset exported successfully!")
```

## Key Observations

1. The Orders, Customers, and Products datasets were successfully combined using the `merge()` function based on `Customer_ID` and `Product_ID`.
2. The `concat()` function was demonstrated by combining selected portions of the Orders DataFrame.
3. The `Order_Date` column was converted into DateTime format, making it possible to extract the order month, day, and day of the week.
4. A new `Total_Amount` column was created by multiplying `Quantity` and `Unit_Price`.
5. The `apply()` function was used to classify orders into Low Value, Medium Value, and High Value categories.
6. Customer membership information was transformed into a more descriptive `Customer_Category` column.
7. The final processed dataset combines order, customer, and product information into a single clean and meaningful DataFrame.
8. The processed dataset contains 120 records and can be used for further sales analysis and reporting.
