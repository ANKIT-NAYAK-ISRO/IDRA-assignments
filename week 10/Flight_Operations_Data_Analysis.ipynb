# Complete Flight Operations Data Analysis Using Python and Pandas

## Objective

The objective of this project is to perform a complete data analysis workflow on a Flight Operations dataset using Python and Pandas. The analysis includes loading and understanding the dataset, inspecting its structure and statistics, handling missing values, selecting and filtering relevant data, sorting records, performing grouping and aggregation, creating useful transformed columns, identifying patterns, and presenting key findings.


## 1. Import Required Library

```python
import pandas as pd
```

## 2. Load the Dataset

```python
df = pd.read_csv("Day10_Flight_Operations_Dataset.csv")
df.head()
```

## 3. Understand the Dataset Structure

```python
print("Dataset Shape:", df.shape)
print("\nColumns:")
print(df.columns.tolist())
print("\nData Types:")
print(df.dtypes)
print("\nDataset Information:")
df.info()
```

## 4. Inspect First, Last and Random Records

```python
print("First 5 Rows:")
display(df.head())

print("Last 5 Rows:")
display(df.tail())

print("Random Sample:")
display(df.sample(5, random_state=42))
```

## 5. Generate Descriptive Statistics

```python
df.describe()
```

## 6. Check Missing Values

```python
print(df.isnull().sum())
```

## 7. Handle Missing Values and Convert Date

```python
# Convert Flight_Date to DateTime
df["Flight_Date"] = pd.to_datetime(df["Flight_Date"])

# Fill missing delay values using the median delay
df["Delay_Minutes"] = df["Delay_Minutes"].fillna(
    df["Delay_Minutes"].median()
)

print("Missing values after cleaning:")
print(df.isnull().sum())
```

## 8. Data Selection

```python
selected_data = df[[
    "Flight_ID",
    "Airline",
    "Origin",
    "Destination",
    "Passengers",
    "Average_Ticket_Price",
    "Delay_Minutes",
    "Flight_Status"
]]

selected_data.head()
```

## 9. Filtering Relevant Flight Records

```python
# Flights with more than 150 passengers
high_passenger_flights = df[df["Passengers"] > 150]
display(high_passenger_flights.head())

# Delayed flights with delays greater than 30 minutes
major_delays = df[
    (df["Flight_Status"] == "Delayed") &
    (df["Delay_Minutes"] > 30)
]
display(major_delays.head())

# Business class flights
business_flights = df[df["Travel_Class"] == "Business"]
display(business_flights.head())
```

## 10. Sorting Operations

```python
# Top 10 flights by passenger count
top_passenger_flights = df.sort_values(
    by="Passengers",
    ascending=False
).head(10)

top_passenger_flights[[
    "Flight_ID",
    "Airline",
    "Passengers",
    "Destination"
]]
```

```python
# Flights with the highest delays
df.sort_values(
    by="Delay_Minutes",
    ascending=False
).head(10)[[
    "Flight_ID",
    "Airline",
    "Delay_Minutes",
    "Weather",
    "Flight_Status"
]]
```

## 11. Create Useful Transformed Columns

```python
# Calculate seat occupancy percentage
df["Occupancy_Rate"] = (
    df["Passengers"] / df["Seat_Capacity"] * 100
).round(2)

# Estimate flight revenue
df["Estimated_Revenue"] = (
    df["Passengers"] * df["Average_Ticket_Price"]
)

# Extract date information
df["Flight_Month"] = df["Flight_Date"].dt.month_name()
df["Day_of_Week"] = df["Flight_Date"].dt.day_name()

# Categorize delays
df["Delay_Category"] = df["Delay_Minutes"].apply(
    lambda x: "On Time"
    if x == 0
    else ("Minor Delay" if x <= 30 else "Major Delay")
)

# Categorize passenger satisfaction
df["Satisfaction_Level"] = df["Passenger_Satisfaction"].apply(
    lambda x: "High"
    if x >= 4
    else ("Medium" if x == 3 else "Low")
)

df.head()
```

## 12. Airline-wise Analysis using groupby() and Aggregation

```python
airline_analysis = df.groupby("Airline").agg(
    Total_Flights=("Flight_ID", "count"),
    Total_Passengers=("Passengers", "sum"),
    Average_Delay=("Delay_Minutes", "mean"),
    Average_Satisfaction=("Passenger_Satisfaction", "mean"),
    Estimated_Revenue=("Estimated_Revenue", "sum"),
    Average_Occupancy=("Occupancy_Rate", "mean")
)

airline_analysis.sort_values(
    by="Estimated_Revenue",
    ascending=False
)
```

## 13. Flight Status Analysis

```python
status_analysis = df.groupby("Flight_Status").agg(
    Total_Flights=("Flight_ID", "count"),
    Average_Delay=("Delay_Minutes", "mean"),
    Average_Satisfaction=("Passenger_Satisfaction", "mean")
)

status_analysis.sort_values(
    by="Total_Flights",
    ascending=False
)
```

## 14. Travel Class Analysis

```python
travel_class_analysis = df.groupby("Travel_Class").agg(
    Total_Flights=("Flight_ID", "count"),
    Total_Passengers=("Passengers", "sum"),
    Average_Ticket_Price=("Average_Ticket_Price", "mean"),
    Estimated_Revenue=("Estimated_Revenue", "sum")
)

travel_class_analysis.sort_values(
    by="Estimated_Revenue",
    ascending=False
)
```

## 15. Weather and Delay Analysis

```python
weather_analysis = df.groupby("Weather").agg(
    Total_Flights=("Flight_ID", "count"),
    Average_Delay=("Delay_Minutes", "mean")
)

weather_analysis.sort_values(
    by="Average_Delay",
    ascending=False
)
```

## 16. Booking Channel Analysis

```python
booking_analysis = df.groupby("Booking_Channel").agg(
    Total_Flights=("Flight_ID", "count"),
    Total_Passengers=("Passengers", "sum"),
    Estimated_Revenue=("Estimated_Revenue", "sum")
)

booking_analysis.sort_values(
    by="Estimated_Revenue",
    ascending=False
)
```

## 17. Top Flights by Estimated Revenue

```python
top_revenue_flights = df.nlargest(
    10,
    "Estimated_Revenue"
)

top_revenue_flights[[
    "Flight_ID",
    "Airline",
    "Travel_Class",
    "Passengers",
    "Average_Ticket_Price",
    "Estimated_Revenue"
]]
```

## Key Observations

1. The Flight Operations dataset contains **180 flight records and 17 original features**.
2. The `Delay_Minutes` column initially contained **7 missing values**, which were handled using the median delay value.
3. **Vistara** generated the highest estimated revenue among the airlines, with approximately **₹28,689,467** in estimated revenue.
4. **Economy Class** accounted for the highest estimated revenue, approximately **₹73,536,524**, and also represented the largest number of flights.
5. Most flights were **On Time (114 flights)**, while **59 flights were delayed** and only **7 flights were cancelled**.
6. Delayed flights had a much higher average delay of **50 minutes** compared with approximately **3.46 minutes** for on-time flights.
7. **Rain** was associated with the highest average delay among the weather conditions, at approximately **21.42 minutes**.
8. **Travel Agency** was the highest revenue-generating booking channel, producing approximately **₹35,413,370** in estimated revenue.
