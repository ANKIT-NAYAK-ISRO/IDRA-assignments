```python
total_sales = df['Total_Sales'].sum()
average_sales = df['Total_Sales'].mean()
highest_sales = df['Total_Sales'].max()
lowest_sales = df['Total_Sales'].min()
total_quantity = df['Quantity'].sum()
number_of_orders = df['Order_ID'].nunique()

print("Total Sales:", round(total_sales, 2))
print("Average Sales:", round(average_sales, 2))
print("Highest Sales:", highest_sales)
print("Lowest Sales:", lowest_sales)
print("Total Quantity Sold:", total_quantity)
print("Number of Orders:", number_of_orders)
```

```python
category_analysis = df.groupby('Category').agg(
    Total_Sales=('Total_Sales', 'sum'),
    Average_Sales=('Total_Sales', 'mean'),
    Highest_Sale=('Total_Sales', 'max'),
    Lowest_Sale=('Total_Sales', 'min'),
    Total_Quantity=('Quantity', 'sum'),
    Number_of_Orders=('Order_ID', 'count')
)

category_analysis = category_analysis.sort_values(
    by='Total_Sales',
    ascending=False
)

category_analysis
```

```python
city_analysis = df.groupby('City').agg(
    Total_Sales=('Total_Sales', 'sum'),
    Average_Sales=('Total_Sales', 'mean'),
    Total_Quantity=('Quantity', 'sum'),
    Number_of_Orders=('Order_ID', 'count')
)

city_analysis = city_analysis.sort_values(
    by='Total_Sales',
    ascending=False
)

city_analysis
```

```python
product_analysis = df.groupby('Product').agg(
    Total_Sales=('Total_Sales', 'sum'),
    Total_Quantity=('Quantity', 'sum'),
    Average_Sales=('Total_Sales', 'mean'),
    Number_of_Orders=('Order_ID', 'count')
)

product_analysis = product_analysis.sort_values(
    by='Total_Sales',
    ascending=False
)

product_analysis.head(10)
```

```python
payment_analysis = df.groupby('Payment_Method').agg(
    Total_Sales=('Total_Sales', 'sum'),
    Average_Sales=('Total_Sales', 'mean'),
    Number_of_Orders=('Order_ID', 'count')
)

payment_analysis = payment_analysis.sort_values(
    by='Total_Sales',
    ascending=False
)

payment_analysis
```
