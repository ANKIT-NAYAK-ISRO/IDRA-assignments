# Cleaned Company Employee Dataset

## Objective

The objective of this assignment is to inspect a messy company employee dataset, identify missing values, inconsistent categorical entries, incorrect data types, and duplicate records, and apply suitable data-cleaning techniques using Pandas. The cleaned dataset is then verified, compared with the original dataset, and exported as a CSV file.


## 1. Import Pandas and Load the Dataset

```python
import pandas as pd

df = pd.read_csv("Day11_Messy_Company_Employee_Dataset.csv")

df.head()
```

## 2. Initial Dataset Inspection

```python
print("Shape:", df.shape)
print("\nColumns:")
print(df.columns.tolist())

print("\nData Types:")
print(df.dtypes)

print("\nDataset Information:")
df.info()
```

## 3. Identify Missing Values

```python
missing_values = df.isnull().sum()
missing_percentage = (df.isnull().mean() * 100).round(2)

missing_report = pd.DataFrame({
    "Missing_Count": missing_values,
    "Missing_Percentage": missing_percentage
})

missing_report[missing_report["Missing_Count"] > 0]
```

### Missing-value findings

The original dataset contains missing values in several fields. Numeric fields such as Age, Annual_Salary, Experience_Years, and Performance_Score are suitable for median imputation because the median is less affected by unusually high or low values. Categorical fields such as Department, Gender, City, and Work_Mode are filled with their mode because these fields represent categories.

`dropna()` is not used for these fields because removing those rows would unnecessarily reduce the dataset. `forward fill` is also not appropriate for employee attributes because the previous employee's value should not be assumed to apply to the next employee.


## 4. Identify Inconsistent Categorical Entries

```python
categorical_columns = [
    "Department", "Gender", "City", "Work_Mode"
]

for column in categorical_columns:
    print(f"\n{column}:")
    print(sorted(df[column].dropna().astype(str).unique()))
```

## 5. Identify Duplicate Records

```python
duplicate_count = df.duplicated().sum()

print("Number of duplicate rows:", duplicate_count)

df[df.duplicated(keep=False)].sort_values("Employee_ID")
```

## 6. Clean Text and Standardize Inconsistent Entries

```python
# Remove leading/trailing whitespace
text_columns = [
    "Employee_Name", "Department", "Job_Title",
    "Gender", "City", "Work_Mode"
]

for column in text_columns:
    df[column] = df[column].astype("string").str.strip()

# Standardize inconsistent Department values
df["Department"] = df["Department"].replace({
    "ENGINEERING": "Engineering",
    "engineering": "Engineering"
})

# Standardize Gender values
df["Gender"] = df["Gender"].replace({
    "MALE": "Male",
    "female": "Female"
})

# Standardize City values
df["City"] = df["City"].replace({
    "DELHI": "Delhi",
    "Delhi ": "Delhi",
    "delhi": "Delhi"
})

# Standardize Work_Mode values
df["Work_Mode"] = df["Work_Mode"].replace({
    "REMOTE": "Remote",
    "remote": "Remote"
})

print("Categorical values standardized.")
```

## 7. Correct Data Types

```python
# Convert numeric columns explicitly
numeric_columns = [
    "Age",
    "Annual_Salary",
    "Experience_Years",
    "Performance_Score"
]

for column in numeric_columns:
    df[column] = pd.to_numeric(df[column], errors="coerce")

# Convert Employee_ID to string
df["Employee_ID"] = df["Employee_ID"].astype("string")

# Convert Joining_Date to datetime
df["Joining_Date"] = pd.to_datetime(
    df["Joining_Date"],
    errors="coerce"
)

df.dtypes
```

## 8. Handle Missing Numeric Values Using Median

```python
for column in [
    "Age",
    "Annual_Salary",
    "Experience_Years",
    "Performance_Score"
]:
    df[column] = df[column].fillna(df[column].median())

print("Missing numeric values after median imputation:")
print(df[
    ["Age", "Annual_Salary", "Experience_Years", "Performance_Score"]
].isnull().sum())
```

## 9. Handle Missing Categorical Values Using Mode

```python
for column in [
    "Department",
    "Gender",
    "City",
    "Work_Mode"
]:
    df[column] = df[column].fillna(df[column].mode()[0])

print("Missing categorical values after mode imputation:")
print(df[
    ["Department", "Gender", "City", "Work_Mode"]
].isnull().sum())
```

## 10. Remove Duplicate Records

```python
rows_before_duplicates = len(df)

df = df.drop_duplicates().reset_index(drop=True)

duplicates_removed = rows_before_duplicates - len(df)

print("Duplicate rows removed:", duplicates_removed)
print("Rows after removing duplicates:", len(df))
```

## 11. Final Verification

```python
print("Final Shape:", df.shape)

print("\nRemaining Missing Values:")
print(df.isnull().sum())

print("\nRemaining Duplicate Rows:", df.duplicated().sum())

print("\nFinal Data Types:")
print(df.dtypes)

print("\nFinal Dataset Preview:")
display(df.head())
```

## 12. Before vs After Cleaning Comparison

```python
before = {
    "Rows": len(pd.read_csv("Day11_Messy_Company_Employee_Dataset.csv")),
    "Columns": len(pd.read_csv("Day11_Messy_Company_Employee_Dataset.csv").columns),
    "Missing_Values": pd.read_csv("Day11_Messy_Company_Employee_Dataset.csv").isnull().sum().sum(),
    "Duplicate_Rows": pd.read_csv("Day11_Messy_Company_Employee_Dataset.csv").duplicated().sum()
}

after = {
    "Rows": len(df),
    "Columns": len(df.columns),
    "Missing_Values": df.isnull().sum().sum(),
    "Duplicate_Rows": df.duplicated().sum()
}

comparison = pd.DataFrame({
    "Before Cleaning": before,
    "After Cleaning": after
})

comparison
```

## 13. Export the Cleaned Dataset

```python
df.to_csv(
    "Cleaned_Company_Employee_Dataset.csv",
    index=False
)

print("Cleaned dataset exported successfully!")
```

## Cleaning Summary

- The original dataset contained **157 rows and 12 columns**.
- **7 duplicate rows** were identified and removed, leaving **150 unique records**.
- Missing numeric values in Age, Annual_Salary, Experience_Years, and Performance_Score were filled using **median imputation**.
- Missing categorical values in Department, Gender, City, and Work_Mode were filled using **mode imputation**.
- Inconsistent text values were standardized, including variations of Engineering, Male/Female, Delhi, and Remote.
- Leading/trailing whitespace was removed from text fields.
- Numeric columns were converted to appropriate numeric data types.
- Joining_Date was converted to the `datetime64` data type.
- `dropna()` was not used because removing employee records would unnecessarily reduce the dataset.
- Forward filling was not used because employee attributes are independent records and should not inherit values from previous rows.
- After cleaning, the dataset contains **150 rows, 12 columns, zero missing values, and zero duplicate rows**.


## Conclusion

The messy company employee dataset was successfully cleaned using appropriate Pandas techniques. Missing values were handled using median and mode imputation, inconsistent categorical values were standardized, data types were corrected, and duplicate records were removed. The before-and-after comparison confirms that the final dataset is more consistent and complete, with no remaining missing values or duplicate records. The cleaned dataset was exported as `Cleaned_Company_Employee_Dataset.csv` for further analysis.
